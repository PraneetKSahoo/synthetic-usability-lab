```python
# Colab has no reachable localhost, so the app needs a public share tunnel here.
# (Locally, leave GRADIO_SHARE unset and just use the printed local URL.)
os.environ['GRADIO_SHARE'] = '1'

# Pass the HF token to the subprocess explicitly: `!python app.py` runs outside
# the notebook kernel, where Colab Secrets (userdata) are unreachable.
try:
    from google.colab import userdata
    tok = userdata.get('HF_TOKEN')
    if tok:
        os.environ['HF_TOKEN'] = tok
except Exception:
    pass  # falls back to the token saved on disk by login() above

!python app.py
```
